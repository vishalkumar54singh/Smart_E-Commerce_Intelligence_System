```python
import pandas as pd
import numpy as np
import os
```

```python
Data_Path = r"D:\Projects\Smart_E-Commerce_Demand_&_Operations_Intelligence_System\01_Raw_Data"

customers = pd.read_csv(Data_Path + r"\olist_customers_dataset.csv")
geolocation = pd.read_csv(Data_Path + r"\olist_geolocation_dataset.csv")
order_items = pd.read_csv(Data_Path + r"\olist_order_items_dataset.csv")
payments = pd.read_csv(Data_Path + r"\olist_order_payments_dataset.csv")
reviews = pd.read_csv(Data_Path + r"\olist_order_reviews_dataset.csv")
orders = pd.read_csv(Data_Path + r"\olist_orders_dataset.csv")
products = pd.read_csv(Data_Path + r"\olist_products_dataset.csv")
sellers = pd.read_csv(Data_Path + r"\olist_sellers_dataset.csv")
category_translation = pd.read_csv(Data_Path + r"\product_category_name_translation.csv")
```

```python
datasets = {
    "customers": customers,
    "orders": orders,
    "order_items": order_items,
    "products": products,
    "sellers": sellers,
    "payments": payments,
    "reviews": reviews,
    "geolocation": geolocation,
    "category_translation": category_translation
}

for name, df in datasets.items():
    print(f"{name:20} -> {df.shape[0]:>7,} rows × {df.shape[1]:>2} columns")
```

```python
for name, df in datasets.items():
    print(f"\n{'=' * 60}")
    print(name.upper())
    print("=" * 60)
    print(list(df.columns))
```

```python
for name, df in datasets.items():
    print(f"\n{name.upper()}")
    print(df.dtypes)
```

```python
for name, df in datasets.items():
    missing = df.isnull().sum()
    missing = missing[missing > 0].sort_values(ascending=False)

    print(f"\n{name.upper()}")

    if missing.empty:
        print("No missing values.")
    else:
        print(missing)
```

```python
for name, df in datasets.items():
    duplicate_count = df.duplicated().sum()
    print(f"{name:20} -> {duplicate_count:,} duplicate rows")
```

```python
id_columns = {
    "customers": ["customer_id", "customer_unique_id"],
    "orders": ["order_id", "customer_id"],
    "order_items": ["order_id", "order_item_id", "product_id", "seller_id"],
    "products": ["product_id"],
    "sellers": ["seller_id"],
    "payments": ["order_id", "payment_sequential"],
    "reviews": ["review_id", "order_id"],
    "category_translation": ["product_category_name"],
}

for dataset_name, columns in id_columns.items():
    df = datasets[dataset_name]

    print(f"\n{dataset_name.upper()}")

    for column in columns:
        if column in df.columns:
            print(f"{column:35} unique values: {df[column].nunique():,}")
```

```python
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"],
    errors="coerce"
)

print("First order:", orders["order_purchase_timestamp"].min())
print("Last order :", orders["order_purchase_timestamp"].max())
```

```python
print("Orders:", orders["order_id"].nunique())
print("Order items:", order_items["order_id"].nunique())
print("Customers:", customers["customer_id"].nunique())
print("Products:", products["product_id"].nunique())
print("Sellers:", sellers["seller_id"].nunique())
```

```python
order_customer_check = orders["customer_id"].isin(
    customers["customer_id"]
)
print(
    "Orders with valid customer_id:",
    f"{order_customer_check.mean() * 100:.2f}%"
)

#To check order_item
order_item_check = order_items["order_id"].isin(
    orders["order_id"]
)
print(
    "Order items with valid order_id:",
    f"{order_item_check.mean() * 100:.2f}%"
)

#To check product
product_check = order_items["product_id"].isin(
    products["product_id"]
)
print(
    "Order items with valid product_id:",
    f"{product_check.mean() * 100:.2f}%"
)

#To check seller
seller_check = order_items["seller_id"].isin(
    sellers["seller_id"]
)
print(
    "Order items with valid seller_id:",
    f"{seller_check.mean() * 100:.2f}%"
)
```

```python
summary = []

for name, df in datasets.items():
    summary.append({
        "dataset": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "missing_cells": int(df.isnull().sum().sum()),
        "duplicate_rows": int(df.duplicated().sum())
    })

summary_df = pd.DataFrame(summary)

summary_df
```
